#### **Array creation and operation on it.**

```python
import numpy as np
```

#### **reshaping array**

```python
import numpy as np

arr4 = np.arange(12)
print("arr4:\n", arr4)

# 1. Reshape into 3 rows, 4 columns (3 * 4 = 12 elements)
reshaped = np.reshape(arr4, (3, 4))
print("\nReshaped (3x4):\n", reshaped)

# 2. Flatten back into 1D array
flattened = reshaped.flatten()
print("\nFlattened:\n", flattened)

# 3. Transpose a 2D array to flip rows and columns
transposed = np.transpose(reshaped)
print("\nTransposed (4x3):\n", transposed)

# 4. Joining operations
a = np.array([2, 4, 6])
b = np.array([8, 10, 12])

print("\nConcatenate:", np.concatenate((a, b)))
print("Vstack:\n", np.vstack((a, b)))
print("Hstack:", np.hstack((a, b)))
```

#### **mathematical operation**

```python
a = np.array([[1, 2, 3],[3,4,5]])
b = np.array([[4, 5, 6],[7,8,9]])

print( a + b)
print( a - b)
print( a * b)
print( a / b)
print( a ** 2)

print(np.sqrt(b))
print(np.exp(a))
print(np.log(a))
print(np.sin(a))
print(np.cos(a))
```

#### **matrix (linear algebra)**

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

print("Matrix multiply (dot):\n", np.dot(a, b))
print("Inverse of A:\n", np.linalg.inv(a))
print("Determinant of A:", np.linalg.det(b))
print("\n")
eigvals, eigvecs = np.linalg.eig(a)
print("Eigenvalues:", eigvals)
print("Eigenvectors:\n", eigvecs)
```
